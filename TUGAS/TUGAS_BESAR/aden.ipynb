```python
import pandas as pd

# Load the datasets
nyc_data_path = '/content/dataset_TSMC2014_NYC.csv'
tokyo_data_path = '/content/dataset_TSMC2014_TKY.csv'

# Read the data
nyc_data = pd.read_csv(nyc_data_path)
tokyo_data = pd.read_csv(tokyo_data_path)

# Display the first few rows and basic info to understand structure
nyc_data_info = nyc_data.info()
nyc_data_preview = nyc_data.head()

tokyo_data_info = tokyo_data.info()
tokyo_data_preview = tokyo_data.head()

nyc_data_info, nyc_data_preview, tokyo_data_info, tokyo_data_preview
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Extracting top venues with the highest number of check-ins for both cities
nyc_top_venues = nyc_data['venueCategory'].value_counts().head(10)
tokyo_top_venues = tokyo_data['venueCategory'].value_counts().head(10)

# Plotting top venues for NYC
plt.figure(figsize=(12, 6))
sns.barplot(x=nyc_top_venues.values, y=nyc_top_venues.index, palette='viridis')
plt.title("Top 10 Venue Categories in New York City", fontsize=16)
plt.xlabel("Number of Check-ins", fontsize=12)
plt.ylabel("Venue Category", fontsize=12)
plt.show()

# Plotting top venues for Tokyo
plt.figure(figsize=(12, 6))
sns.barplot(x=tokyo_top_venues.values, y=tokyo_top_venues.index, palette='plasma')
plt.title("Top 10 Venue Categories in Tokyo", fontsize=16)
plt.xlabel("Number of Check-ins", fontsize=12)
plt.ylabel("Venue Category", fontsize=12)
plt.show()

nyc_top_venues, tokyo_top_venues
```

```python
# Plotting scatter plots for location density
plt.figure(figsize=(12, 8))
sns.scatterplot(data=nyc_data, x='longitude', y='latitude', alpha=0.3, color='blue')
plt.title('Location Density of Check-ins in New York City', fontsize=16)
plt.xlabel('Longitude', fontsize=12)
plt.ylabel('Latitude', fontsize=12)
plt.grid(True)
plt.show()

plt.figure(figsize=(12, 8))
sns.scatterplot(data=tokyo_data, x='longitude', y='latitude', alpha=0.3, color='red')
plt.title('Location Density of Check-ins in Tokyo', fontsize=16)
plt.xlabel('Longitude', fontsize=12)
plt.ylabel('Latitude', fontsize=12)
plt.grid(True)
plt.show()
```

```python
# Parsing timestamp and extracting seasonal data
nyc_data['utcTimestamp'] = pd.to_datetime(nyc_data['utcTimestamp'])
tokyo_data['utcTimestamp'] = pd.to_datetime(tokyo_data['utcTimestamp'])

# Function to assign seasons
def get_season(month):
    if month in [12, 1, 2]:
        return 'Winter'
    elif month in [3, 4, 5]:
        return 'Spring'
    elif month in [6, 7, 8]:
        return 'Summer'
    elif month in [9, 10, 11]:
        return 'Autumn'

# Adding season column
nyc_data['Season'] = nyc_data['utcTimestamp'].dt.month.apply(get_season)
tokyo_data['Season'] = tokyo_data['utcTimestamp'].dt.month.apply(get_season)

# Count check-ins by season
nyc_season_counts = nyc_data['Season'].value_counts()
tokyo_season_counts = tokyo_data['Season'].value_counts()

# Plot seasonal activity
plt.figure(figsize=(12, 6))
sns.barplot(x=nyc_season_counts.index, y=nyc_season_counts.values, palette='cool')
plt.title("Seasonal Check-ins in New York City", fontsize=16)
plt.xlabel("Season", fontsize=12)
plt.ylabel("Number of Check-ins", fontsize=12)
plt.show()

plt.figure(figsize=(12, 6))
sns.barplot(x=tokyo_season_counts.index, y=tokyo_season_counts.values, palette='autumn')
plt.title("Seasonal Check-ins in Tokyo", fontsize=16)
plt.xlabel("Season", fontsize=12)
plt.ylabel("Number of Check-ins", fontsize=12)
plt.show()

nyc_season_counts, tokyo_season_counts
```

```python
# Filter for indoor venues in NYC
indoor_categories_nyc = ['Shopping Mall', 'Theater', 'Museum', 'Gym / Fitness Center']
nyc_indoor = nyc_data[nyc_data['venueCategory'].isin(indoor_categories_nyc)]

# Count check-ins by season for indoor venues
nyc_indoor_season_counts = nyc_indoor['Season'].value_counts().sort_index()

# Plot seasonal activity for indoor locations in NYC
plt.figure(figsize=(12, 6))
sns.barplot(x=nyc_indoor_season_counts.index, y=nyc_indoor_season_counts.values, palette='cool')
plt.title("Seasonal Check-ins at Indoor Locations in New York", fontsize=16)
plt.xlabel("Season", fontsize=12)
plt.ylabel("Number of Check-ins", fontsize=12)
plt.xticks(fontsize=10)
plt.yticks(fontsize=10)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()

nyc_indoor_season_counts
```
